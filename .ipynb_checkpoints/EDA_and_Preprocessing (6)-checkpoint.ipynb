## Этап 2. Предварительная обработка данных

### 2.1 Работа с пропусками
Удалим явные дубликаты и строки с логическими аномалиями.

```python
df = df.drop_duplicates().reset_index(drop=True)
df = df[(df['total_area'] >= 12) & (df['build_year'] >= 1850) & (df['build_year'] <= 2026) & (df['floor'] <= df['floors_total'])].copy()
```

Заполним пропуски в категориальном признаке `house_type` модой по району, а пропуски в жилой и кухонной площади - медианной долей от общей площади.

```python
df['house_type'] = df.groupby('district')['house_type'].transform(lambda x: x.fillna(x.mode()[0] if not x.mode().empty else 'Панельный'))

living_ratio = (df['living_area'] / df['total_area']).median()
df['living_area'] = df['living_area'].fillna(df['total_area'] * living_ratio)

kitchen_ratio = (df['kitchen_area'] / df['total_area']).median()
df['kitchen_area'] = df['kitchen_area'].fillna(df['total_area'] * kitchen_ratio)
```

### 2.2 Обработка выбросов и логарифмирование
Применим логарифмирование целевой переменной `price` для снижения асимметрии.

```python
df['log_price'] = np.log1p(df['price'])
print(f"Асимметрия до: {skew(df['price']):.2f}, после логарифмирования: {skew(df['log_price']):.2f}")
```

### 2.3 Feature Engineering
Сгенерируем новые производные признаки для улучшения качества будущих моделей.

```python
df['building_age'] = 2026 - df['build_year']
df['price_per_sqm'] = df['price'] / df['total_area']
df['living_ratio'] = df['living_area'] / df['total_area']
df['kitchen_ratio'] = df['kitchen_area'] / df['total_area']

df['is_first_floor'] = (df['floor'] == 1).astype(int)
df['is_last_floor'] = (df['floor'] == df['floors_total']).astype(int)
df['floor_category'] = df.apply(lambda r: 'Первый' if r['floor'] == 1 else ('Последний' if r['floor'] == r['floors_total'] else 'Средний'), axis=1)
```

### 2.4 Кодирование категориальных признаков
Закодируем все категориальные столбцы с помощью метода One-Hot Encoding.

```python
df_encoded = pd.get_dummies(df, columns=['district', 'house_type', 'renovation', 'floor_category'], drop_first=True, dtype=int)
df_encoded.head()
```

Удалим неинформативные и избыточные колонки (`id`, `build_year`, `price_per_sqm`).

```python
drop_cols = ['id', 'build_year', 'price_per_sqm']
df_prepared = df_encoded.drop(columns=[c for c in drop_cols if c in df_encoded.columns])
```

### 2.5 Разбиение данных
Разделим выборку на обучающий (train) и тестовый (test) наборы в пропорции 80/20 с фиксированным `random_state=42`.

```python
X = df_prepared.drop(columns=['price', 'log_price'])
y = df_prepared['log_price']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
print(f"Формат X_train: {X_train.shape}, X_test: {X_test.shape}")
```

### 2.6 Масштабирование признаков
Применим `StandardScaler` к числовым колонкам, обучив скалер исключительно на `X_train` для исключения утечки данных (data leakage).

```python
num_features = ['rooms', 'total_area', 'living_area', 'kitchen_area', 'floor', 'floors_total', 'distance_to_center', 'building_age', 'living_ratio', 'kitchen_ratio']

scaler = StandardScaler()
X_train_scaled = X_train.copy()
X_test_scaled = X_test.copy()

X_train_scaled[num_features] = scaler.fit_transform(X_train[num_features])
X_test_scaled[num_features] = scaler.transform(X_test[num_features])
```

Сохраним подготовленные обработанные данные в отдельный CSV-файл.

```python
df_prepared.to_csv('cleaned_real_estate.csv', index=False)
```

**Вывод по этапам 1 и 2:**
1. Выполнен полный статистический и визуальный анализ набора данных.
2. Удалены дубликаты и аномалии, корректно заполнены пропущенные значения.
3. Сгенерированы новые признаки, проведено кодирование One-Hot Encoding.
4. Выборки разделены без утечки данных, сформирован готовый датасет `cleaned_real_estate.csv`.
