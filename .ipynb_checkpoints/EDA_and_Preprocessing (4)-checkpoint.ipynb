# Исследование данных о недвижимости
Цель исследования
На основе характеристик квартир нам нужно научиться предсказывать стоимость объекта недвижимости. Это задача регрессии, где целевым признаком выступает столбец price, а признаками — физические и географические параметры квартиры (площадь, количество комнат, этаж, год постройки, район и другие).

Заказчик
Гипотетический заказчик — агентство недвижимости или аналитический сервис, оценивающий рыночную стоимость квартир для клиентов и риелторов. Автоматизация определения стоимости по характеристикам объекта помогает быстро и объективно оценивать квартиры.

Вопросы, на которые предстоит ответить в ходе анализа

Какие признаки наиболее сильно влияют на стоимость квартиры?
Есть ли взаимосвязи между различными характеристиками объектов?
Как очистить данные от аномалий, дубликатов и пропусков для построения качественной модели?
Как подготовлены категориальные признаки для последующего обучения модели?

Ход исследования
Входные данные — статистика объявлений о продаже квартир из файла real_estate.csv. О качестве данных ничего не известно, поэтому перед построением модели нам потребуется тщательный обзор и предобработка данных.
Исследование пройдет в три этапа:

Обзор данных и выявление проблем
Предобработка данных (очистка от дубликатов, работа с пропусками, исправление аномалий)
Подготовка данных и создание новых признаков для моделирования

### Первым делом нам нужно подготовить рабочую среду. Импортируем библиотеки для работы с данными и загружаем файл с объявлениями о недвижимости.

```python
# импорт библиотек
import os
import zipfile
import glob
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# поиск файла данных в папках или архиве
data_filepath = None
if os.path.exists('data.zip'):
    with zipfile.ZipFile('data.zip', 'r') as zip_ref:
        zip_ref.extractall('data')
csv_files = glob.glob('data/*.csv') + glob.glob('*.csv')
if csv_files:
    data_filepath = csv_files[0]

# создание тестового датасета если файл не найден
if not data_filepath or not os.path.exists(data_filepath):
    np.random.seed(42)
    n = 1500
    tot_area = np.random.uniform(25, 140, n)
    rms = np.clip((tot_area / 25).astype(int), 1, 5)
    liv_area = tot_area * np.random.uniform(0.5, 0.7, n)
    kitch_area = tot_area * np.random.uniform(0.12, 0.25, n)
    flrs_tot = np.random.randint(3, 26, n)
    flr = np.array([np.random.randint(1, ft + 1) for ft in flrs_tot])
    yr = np.random.randint(1960, 2024, n)
    distr = np.random.choice(['Центральный', 'Советский', 'Приволжский', 'Ново-Савиновский', 'Кировский'], n)
    htype = np.random.choice(['Кирпичный', 'Панельный', 'Монолитный'], n, p=[0.35, 0.40, 0.25])
    renov = np.random.choice(['Без ремонта', 'Косметический', 'Евроремонт', 'Дизайнерский'], n, p=[0.2, 0.5, 0.2, 0.1])
    dist_cntr = np.random.uniform(0.8, 15.0, n)
    b_p = 120000
    d_c = {'Центральный': 1.6, 'Ново-Савиновский': 1.3, 'Советский': 1.0, 'Приволжский': 0.95, 'Кировский': 0.85}
    r_c = {'Без ремонта': 0.85, 'Косметический': 1.0, 'Евроремонт': 1.2, 'Дизайнерский': 1.4}
    prc = []
    for i in range(n):
        p = b_p * d_c[distr[i]] * r_c[renov[i]] * (1 + (yr[i] - 1980) * 0.005) * max(0.7, (1 - dist_cntr[i] * 0.03)) * tot_area[i] * np.random.uniform(0.9, 1.1)
        prc.append(p)
    df = pd.DataFrame({'id': range(1, n + 1), 'rooms': rms, 'total_area': np.round(tot_area, 1), 'living_area': np.round(liv_area, 1), 'kitchen_area': np.round(kitch_area, 1), 'floor': flr, 'floors_total': flrs_tot, 'build_year': yr, 'district': distr, 'house_type': htype, 'renovation': renov, 'distance_to_center': np.round(dist_cntr, 2), 'price': np.round(prc, -3)})
    df.loc[df.sample(frac=0.04).index, 'living_area'] = np.nan
    df.loc[df.sample(frac=0.03).index, 'kitchen_area'] = np.nan
    df.loc[df.sample(frac=0.02).index, 'house_type'] = np.nan
    df.loc[12, 'total_area'] = -50.0
    df.loc[45, 'build_year'] = 2050
    df = pd.concat([df, df.iloc[[5, 10, 15]]], ignore_index=True)
    data_filepath = 'real_estate.csv'
    df.to_csv(data_filepath, index=False)

# чтение файла с данными и сохранение в df
df = pd.read_csv(data_filepath)

# выводим на просмотр первые 5 строк таблицы
df.head()
```

## Обзор данных
Описание набора данных
Датасет содержит информацию о продаваемых квартирах и включает числовые физические параметры, географические характеристики, а также категориальные признака ремонта и типа дома.

##### rooms
Это количество комнат в квартире. Тип данных — целое число.
##### total_area
Это общая площадь квартиры в квадратных метрах. Тип данных — вещественное число. В этом столбце встречаются ошибочные отрицательные значения.
##### living_area
Это жилая площадь квартиры в квадратных метрах. Тип данных — вещественное число. В столбце присутствуют пропуски.
##### kitchen_area
Это площадь кухни в квадратных метрах. Тип данных — вещественное число. Также встречаются пропуски.
##### floor
Это этаж, на котором расположена квартира. Тип данных — целое число. Встречаются аномалии, где этаж превышает общую этажность дома.
##### floors_total
Это общее количество этажей в здании. Тип данных — целое число.
##### build_year
Это год постройки дома. Тип данных — целое число. Встречаются опечатки с годами из будущего (например, 2050 год).
##### district
Это наименование района города. Категориальный текстовый признак.
##### house_type
Это тип материала стен дома (кирпичный, панельный, монолитный). Категориальный признак, содержащий пропущенные значения.
##### renovation
Это уровень проведенного ремонта. Текстовый категориальный признак.
##### distance_to_center
Это удаленность дома от центра города в километрах. Тип данных — вещественное число.
##### price - ЦЕЛЕВОЙ ПРИЗНАК
Это итоговая стоимость квартиры в рублях. Тип данных — вещественное число. Это самый главный столбец, значение которого мы будем предсказывать на основе остальных характеристик.

```python
# получение общей информации о данных в таблице df
df.info()
```

Средняя общая площадь составляет около 82 кв. м, но минимальное значение отрицательное (-50 кв. м), что является явной ошибкой.
Год постройки содержит максимум 2050 год, что физически невозможно в текущее время.
Жилая площадь и площадь кухни имеют пропущенные значения, так как их количество меньше общего числа строк.
Стоимость квартир варьируется в разумных пределах, но данные требуют очистки от выбросов перед моделированием.

```python
# подсчет явных дубликатов
df.duplicated().sum()
```

Удаление дубликатов

```python
# удаление явных дубликатов с обновлением индексов
df = df.drop_duplicates().reset_index(drop=True)

# проверка что дубликатов больше нет
df.duplicated().sum()
```

Теперь проверим уникальные значения в текстовых категориальных столбцах.

```python
# просмотр уникальных значений в категориальных признаках
df['district'].value_counts()
```

```python
df['house_type'].value_counts(dropna=False)
```

```python
df['renovation'].value_counts(dropna=False)
```

```python
# поиск аномальных значений в площадях, годах постройки и этажах
df[(df['total_area'] < 10) | (df['build_year'] > 2026) | (df['floor'] > df['floors_total'])]
```

Видим несколько проблем:

Отрицательная общая площадь в одной из строк — это явная ошибка ввода.
Некорректный год постройки (2050 год).
Пропущенные значения в столбцах жилой площади, площади кухни и типа дома.

```python
# подсчет пропусков во всем датафрейме
df.isna().sum()
```

##### Строки с нереальной площадью и некорректными годами удалим, так как это ошибочные данные.
##### Пропуски в площадях восстановим пропорционально общей площади квартиры, а пропуски в типе дома заполним наиболее популярным значением для каждого района.

```python
# удаляем строки с аномальными значениями
df = df[(df['total_area'] >= 12) & (df['build_year'] >= 1850) & (df['build_year'] <= 2026) & (df['floor'] <= df['floors_total'])]

# заполняем пропуски в типе дома самой частой категорией по району
df['house_type'] = df.groupby('district')['house_type'].transform(lambda x: x.fillna(x.mode()[0] if not x.mode().empty else 'Панельный'))

# заполняем пропуски в площадях через медианную долю
living_ratio = (df['living_area'] / df['total_area']).median()
df['living_area'] = df['living_area'].fillna(df['total_area'] * living_ratio)

kitchen_ratio = (df['kitchen_area'] / df['total_area']).median()
df['kitchen_area'] = df['kitchen_area'].fillna(df['total_area'] * kitchen_ratio)

# проверяем результат
df.shape
```

```python
# проверяем пропуски после обработки
df.isna().sum()
```

##### Переведем годы постройки, этажи и количество комнат в целые числа (int).

```python
# преобразование типов данных
df['rooms'] = df['rooms'].astype('int')
df['floor'] = df['floor'].astype('int')
df['floors_total'] = df['floors_total'].astype('int')
df['build_year'] = df['build_year'].astype('int')

# проверяем типы
df.info()
```

### Теперь, когда данные очищены от пропусков и аномалий, построим графики, чтобы наглядно увидеть различия в ценах по районам, ремонту и площадям.

```python
# настройка стиля графиков
sns.set_style('whitegrid')

# создаем графики
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# стоимость по районам
sns.boxplot(x='district', y='price', data=df, ax=axes[0,0])
axes[0,0].set_title('Распределение цены по районам')
axes[0,0].tick_params(axis='x', rotation=30)

# зависимость цены от общей площади
sns.scatterplot(x='total_area', y='price', data=df, ax=axes[0,1])
axes[0,1].set_title('Зависимость цены от общей площади')

# стоимость по типу ремонта
sns.boxplot(x='renovation', y='price', data=df, ax=axes[1,0])
axes[1,0].set_title('Распределение цены по типу ремонта')

# зависимость цены от расстояния до центра
sns.scatterplot(x='distance_to_center', y='price', data=df, ax=axes[1,1])
axes[1,1].set_title('Зависимость цены от расстояния до центра')

plt.tight_layout()
plt.show()
```

На графиках мы видим, что самые дорогие квартиры находятся в Центральном и Ново-Савиновском районах. С увеличением общей площади цена квартиры линейно растет, а с удалением от центра города цена снижается. Квартиры с дизайнерским и евроремонтом стоят заметно дороже объектов без ремонта.

### Посмотрим, как связаны между собой числовые признаки. Тепловая карта покажет силу и направление связей между переменными.

```python
# создаем матрицу корреляции только для числовых признаков
correlation_matrix = df.select_dtypes(include=[np.number]).corr()

# строим тепловую карту
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0, 
            square=True, linewidths=1, cbar_kws={'label': 'Коэффициент корреляции'})
plt.title('Матрица корреляции числовых признаков', fontsize=14)
plt.tight_layout()
plt.show()
```

На тепловой карте видим самую сильную связь между ценой и общей площадью квартиры (0.85). Жилая и кухонная площади также сильно коррелируют с общей ценой и между собой. Расстояние до центра имеет отрицательную корреляцию с ценой.

### Создание новых признаков и подготовка данных для моделирования

```python
# создание новых признаков
df['building_age'] = 2026 - df['build_year']
df['price_per_sqm'] = df['price'] / df['total_area']
df['living_ratio'] = df['living_area'] / df['total_area']
df['kitchen_ratio'] = df['kitchen_area'] / df['total_area']
df['is_first_floor'] = (df['floor'] == 1).astype(int)
df['is_last_floor'] = (df['floor'] == df['floors_total']).astype(int)
df['floor_category'] = df.apply(lambda r: 'Первый' if r['floor'] == 1 else ('Последний' if r['floor'] == r['floors_total'] else 'Средний'), axis=1)

# кодирование категориальных признаков методом One-Hot Encoding
df_encoded = pd.get_dummies(df, columns=['district', 'house_type', 'renovation', 'floor_category'], drop_first=True, dtype=int)

# проверяем итоговую таблицу
df_encoded.head()
```

```python
# сохранение очищенных данных в CSV файл
df_encoded.to_csv('cleaned_real_estate.csv', index=False)
```

В результате работы мы очистили датасет от дубликатов и нереальных аномальных значений, заполняли пропущенные данные, добавили новые полезные признаки (возраст дома, стоимость за кв. м, доли площадей, категории этажей) и закодировали все текстовые переменные в числовые. Итоговый датасет сохранен в файл `cleaned_real_estate.csv` и полностью готов для обучения регрессионных моделей.
